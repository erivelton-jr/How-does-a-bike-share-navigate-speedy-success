### Process 

For this project I'm going to use Python and Jupyter Notebook, because is a proper language and it's good for documentation steps of the analysis.

```python
# import libraries
import requests
import pandas as pd 
import json
import matplotlib.pyplot as plt
```

```python
# read each file, transform into a dataset and put inside of a list
dfs = []

for file in range(1,13):
    filepath = f'data/raw/2025{file:02}-divvy-tripdata.csv' # data source
    try:
        df_raw = pd.read_csv(filepath, delimiter=',')
        if df_raw.empty:
            print("WARNING: The Dataframe is empty")
        else:
            dfs.append(df_raw)
    except Exception as e:
        print(e)
```

```python
# Display each dataset
for i in range(len(dfs)):
    display(dfs[i].head())
```

```python
# I'm going to use the first dataset columns as pattern for all data
dfs[0].columns
```

So let's check if all the dataset is followng the pattern

```python
for col in range(len(dfs)):
    print(f"==== Dataset {col} ====")
    pattern_col = list(dfs[0].columns)
    comparison_col = list(dfs[col].columns)
    for i in range(len(pattern_col)):
        if comparison_col[i] == pattern_col[i]: 
            continue
        else:
            print(f"The columns {i} is different: \n Col from {dfs[col]}{comparison_col[i]} \n Pattern col: {pattern_col[i]}")
    print("All columns are equal!")
```

It's all good in here!
So now let's concat all data.

```python
df_concat = pd.concat(dfs, ignore_index=True)
df_concat.head()
```

```python
df_concat.describe()
```

```python
df_concat.info()
```

```python
# transforming datetime columns 
df_concat['started_at'] = pd.to_datetime(df_concat['started_at'], format="%Y-%m-%d %H:%M:%S.%f")
df_concat['ended_at'] = pd.to_datetime(df_concat['ended_at'], format="%Y-%m-%d %H:%M:%S.%f")

print(f"started_at: {df_concat['started_at'].dtypes}")
print(f"ended_at: {df_concat['ended_at'].dtypes}")
```

```python
# creating column 'trip_duration'
df_concat['trip_duration_seconds'] = (df_concat['ended_at'] - df_concat['started_at']).dt.total_seconds().astype(int)

# creating column 'day of week'
df_concat['day_of_week'] = df_concat['started_at'].dt.day_name()

# creating each column for date 
df_concat['date'] = df_concat['started_at'].dt.date
df_concat['day'] = df_concat['started_at'].dt.day
df_concat['month'] = df_concat['started_at'].dt.month
df_concat['year'] = df_concat['started_at'].dt.year

df_concat.head()
```

```python
# Checking data info
df_concat.info()
```

```python
# Checking if have duplicated rides
df_concat['ride_id'].duplicated().sum()
```

```python
#Checking unique values
df_concat.nunique()
```

```python
#Checking for null values

df_concat.isna().sum()
```

Now let's remove all the bad data.

```python
# drop null values

df_concat.dropna()
```

```python
df_clean = df_concat[df_concat['trip_duration_seconds'] > 0].copy()
df_clean
```

```python
# Checking the new dataset
df_clean.info()
```

### Analyze 

In this step we will get some answers from the data making some aggregations, calculations and indetifying some trends and relationships.

```python
# Descriptive analysis 
df_clean.describe()
```

```python
# Compare members and casual users
df_clean.groupby('member_casual')['trip_duration_seconds'].agg(['mean', 'median', 'max', 'min'])
```

We can see from that, that casual riders uses the bikes for more time

```python
# Order the days of the week for correct sorting for analysis
days_order = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
df_clean['day_of_week'] = pd.Categorical(df_clean['day_of_week'],
categories=days_order, ordered=True)
```

```python
# See the average ride time by each day for members vs casual users
df_clean.groupby(['member_casual', 'day_of_week'], observed=False)['trip_duration_seconds'].mean()
```

Both of the classes uses bike for more time during the weekend's, but even so, casual riders uses for more time during all week including weekends.

```python
# Analyze ridership data by type and weekday
summary_stats = df_clean.groupby(['member_casual', 'day_of_week'], observed=False).agg(
number_of_rides=('ride_id', 'count'),
average_duration=('trip_duration_seconds', 'mean')
).reset_index()

summary_stats
```
